````python
df_any_dropped = df.dropna()
original_size = df.shape
any_dropped_size = df_any_dropped.shape
data_loss_pct = ((original_size[0] - any_dropped_size[0]) / original_size[0]) * 100

print(f"Исходный размер: {original_size}")
print(f"Новый размер: {any_dropped_size}")
print(f"Потеря данных: {data_loss_pct:.2f}%\n")
# --------------------------------------------------
min_valid_cols = int((1 - 0.3) * len(df))
missing_pcts = df.isna().mean()
dropped_columns_list = missing_pcts[missing_pcts > 0.3].index.tolist()

df_columns_cleaned = df.dropna(axis=1, thresh=min_valid_cols)
print(f"Удалено столбцов: {len(dropped_columns_list)}")
print(f"Удалённые столбцы: {dropped_columns_list}")
print(f"Новый размер: {df_columns_cleaned.shape}\n")
# --------------------------------------------------
threshold_cols = int(len(df.columns) * 0.5)
df_rows_conditional = df_columns_cleaned.dropna(axis=0, thresh=threshold_cols)

print(f"Исходный размер перед удалением строк: {df_columns_cleaned.shape}")
print(f"Размер после удаления строк (> 50% пропусков): {df_rows_conditional.shape}\n")
# --------------------------------------------------
critical_cols = ['employee_id', 'full_name', 'department', 'position', 'salary']
df_final_cleaned = df_rows_conditional.dropna(subset=critical_cols)

print(f"Критические столбцы: {critical_cols}")
print(f"Финальный размер датафрейма: {df_final_cleaned.shape}")
````

Вопросы:
1. Какой метод удаления приводит к наименьшей потере данных?
2. Почему удаление столбцов может быть опасным?
3. В каких случаях стоит использовать стратегическое удаление?

Ответы:
1. Условное удаление строк по порогу. Такой подход минимизирует потерю данных, так как это оставляет некритических и разбросанных пропусков. Вместо того, чтобы выкинуть всю строку из-за отсутствующего номера телефона или и-мейла, удаление по порогу сохраняет их, если они содержат валидные и критические свойства (как один из столбцов заданных выше). Можно также упоминать удаления по критическим столбцам, причина одинаковая.
2. Даже если столбец имеет много пропусков, неосторожное удаление столбцов может привести к исключению критических данных из статистического анализа, смещению выборки и скрытых закономерностей.
3. Стратегическое удаление стоит использовать, если:
   - второстепенный столбец пуст более чем на 30%-50%, а его удаление спасает остальные данные от уничтожения;
   - анализ не имеет смысла без критических свойств;
   - пропуски сгруппированы в одних и тех же строках, стратегическое удаление очищает "непригодные" строки, оставляя основную часть датасета нетронутой.

### Заполнение пропусков

Заполнение константами для разных типов данных

1. Заполнение константами:

*Для категориальных данных* - **"Не указано"**

*Для числовых данных* - **-1**

*Для дат* - **'1900-01-01'**

2. Заполнение статистическими показателями:

*Для каждого числового столбца:*

Сильно асимметричное распределение - медиана

Более симметричное распределение - среднее

*Для категориальных - модой*

3. Групповое заполнение

Зарплата по позиции и городу

Бонус как процент от зарплаты


````python
import numpy as np
import pandas as pd

# "Некорректный номер". Можно его так-то оставить, всё, что не упомянуто в задании, по рассмотрению выполняющего
df_final_cleaned['phone'] = df_final_cleaned['phone'].replace('некорректный номер', np.nan)
# --------------------------------------------------
df_imputed = df_final_cleaned.copy()

numeric_cols = df_imputed.select_dtypes(include=[np.number]).columns.tolist()
categorical_cols = df_imputed.select_dtypes(exclude=[np.number]).columns.tolist()
date_cols = [
    col for col in categorical_cols
    if len(df_imputed[col].dropna()) > 0 and
    ((pd.to_datetime(df_imputed[col].dropna(), errors='coerce', format='mixed').notna().sum() / len(df_imputed[col].dropna())) > 0.8)
]
categorical_cols = [col for col in categorical_cols if col not in date_cols]

imputation_rules = {
    **{col: -1 for col in numeric_cols},
    **{col: 'Не указано' for col in categorical_cols},
    **{col: '1900-01-01' for col in date_cols}
}
df_imputed = df_imputed.fillna(value=imputation_rules)
# --------------------------------------------------
df_stat = df_final_cleaned.copy()

if 'salary' in df_stat.columns and 'position' in df_stat.columns and 'city' in df_stat.columns:
    group_salaries = df_stat.groupby(['position', 'city'])['salary'].transform('median')
    df_stat['salary'] = df_stat['salary'].fillna(group_salaries).fillna(df_stat['salary'].median())

if 'bonus' in df_stat.columns and 'salary' in df_stat.columns and 'department' in df_stat.columns:
    # replace(0, np.nan) чтобы не делить на 0
    bonus_ratios = df_stat['bonus'] / df_stat['salary'].replace(0, np.nan)
    dept_ratio = bonus_ratios.groupby(df_stat['department']).transform('median').fillna(0)
    df_stat['bonus'] = df_stat['bonus'].fillna(df_stat['salary'] * dept_ratio)

exclude_cols = ['salary', 'bonus']
numeric_stats = [c for c in numeric_cols if c not in exclude_cols]

stat_imputation_rules = {
    **{col: (df_stat[col].mean() if abs(df_stat[col].skew()) < 0.5 else df_stat[col].median()) for col in numeric_stats},
    **{col: df_stat[col].mode()[0] for col in categorical_cols if not df_stat[col].mode().empty},
    **{col: '1900-01-01' for col in date_cols}
}
df_stat = df_stat.fillna(value=stat_imputation_rules)

# чтобы как раз подтвердить эффективность заполнения
print(f"NaN-ы в датафрейме с константами: {df_imputed.isna().sum().sum()}")
print(f"NaN-ы в датафрейме с статистиками: {df_stat.isna().sum().sum()}")
````

Вопросы:
1. Как выбрать между медианой и средним для заполнения?
2. Почему групповое заполнение часто лучше общего?
3. Какие риски у заполнения константами?

Ответы:
1. В зависимости от того, как данные распределены в выборке. Если выборка имеет симметричное распределение, то лучше использовать среднюю (там будет большинство, поэтому вероятность того, что пропущенная данная будет иметь такое же значение, достаточно высока). Если выборка имеет асимметричное распределение, то имеет смысл использовать медиану, так как большое неравенство будет влиять на результат среднего. Если представить, что в компании (n=10) 9 сотрудники зарабатывают 100 тысячей рублей каждый, а директор зарабатывает 1,5 миллионов, то средняя зарплата станет 240 тыс., что вдвое больше зарплаты работника. Поэтому при использовании медианы как раз будет использоваться самое вероятное значение.
2. Групповое заполнение учитывает внутренний контекст и взаимосвязи между признаками, что делает заполнение более точным. Такого нет в общем заполнении, где сглаживается различия и уничтожает специфику данных.
3. Такой тип заполнения привлекает за собой следующие риски:
   - Искажение распределения и статистических метрик: если в столбце много пропусков и заполнить их `-1`, то среднее и медиана искусственно смещается вниз, тем самым пагубно влияя на дальнейший анализ;
   - Ложные закономерности: не исключено, что алгоритмы поиска могут считать совершенно разных людей одним и тем же человеком только из-за того, что оба люди якобы родились в `1900-01-01`;
   - Проблемы с алгоритмами МО: некоторые модели будут воспринимать эти значения как действительными и истинными. Модель будет думать, что у сотрудника 126 лет, или что у сотрудника телефон номера `-1`.

# № 3. Обработка дубликатов

Дубликаты могут быть 4-х типов:
*   Полные дубликаты
*   По employee_id
*   По ФИО и email
*   По основным атрибутам ('full_name', 'birth_date', 'city')

Стратегии обработки дубликатов
1. УДАЛЕНИЕ:
* Оставить первую запись
* Оставить последнюю запись
* Удалить все дубликаты

2. Удаление на основе дополнительных условий (например по дате найма)

````python
# В дальнейшем будет использоваться датафрейм со статистическими заполнениями.
duplicate_types = {
    '1. Полные дубликаты': None,
    '2. По employee_id': ['employee_id'],
    '3. По ФИО и email': ['full_name', 'email'],
    '4. По основным атрибутам': ['full_name', 'birth_date', 'city']
}

for title, subset in duplicate_types.items():
    count = df_stat.duplicated(subset=subset).sum()
    print(f"{title}: {count} строк-дубликатов")
print()
# --------------------------------------------------

df_keep_first = df_stat.drop_duplicates(subset=['employee_id'], keep='first')
df_keep_last = df_stat.drop_duplicates(subset=['employee_id'], keep='last')
df_drop_all = df_stat.drop_duplicates(subset=['employee_id'], keep=False)

# --------------------------------------------------
df_sorted_by_hire = df_stat.sort_values(by='hire_date', ascending=True)
df_conditional_clean = df_sorted_by_hire.drop_duplicates(subset=['full_name', 'email'], keep='first')

print(f"Исходный размер до очистки: {df_stat.shape}")
print(f"Размер при оставлении первой записи: {df_keep_first.shape}")
print(f"Размер при оставлении последней записи: {df_keep_last.shape}")
print(f"Размер при удалении всех дубликатов: {df_drop_all.shape}\n")
print(f"Размер финального датасета: {df_conditional_clean.shape}")
````

Ответы:

````python
clean = df.copy()

for column in ['district', 'category', 'priority', 'resolved', 'operator']:
    clean[column] = clean[column].astype('string').str.strip()

clean['category'] = clean['category'].str.lower().replace({
    'lightng': 'lighting',
    'road ': 'road',
    ' road': 'road',
})
clean['priority'] = clean['priority'].str.lower().replace({'': pd.NA})
clean['resolved'] = clean['resolved'].replace({'да': 'yes'})
clean['district'] = clean['district'].replace({'': pd.NA})

for column in ['category', 'priority', 'resolved']:
    print(f'\n{column}:', sorted(clean[column].dropna().astype(str).unique()))
````

## Пропуски и типы данных

````python
clean['created_at'] = pd.to_datetime(clean['created_at'], errors='coerce')
clean['response_hours'] = pd.to_numeric(clean['response_hours'], errors='coerce')
clean['satisfaction_score'] = pd.to_numeric(clean['satisfaction_score'], errors='coerce')
clean['cost_rub'] = pd.to_numeric(clean['cost_rub'], errors='coerce')

clean['district'] = clean['district'].fillna('unknown')
clean['response_hours'] = clean['response_hours'].fillna(clean['response_hours'].median())
clean['satisfaction_score'] = clean['satisfaction_score'].fillna(clean['satisfaction_score'].median())
clean['cost_rub'] = clean['cost_rub'].fillna(clean['cost_rub'].median())
clean['priority'] = clean['priority'].fillna(clean['priority'].mode()[0])

print(clean.dtypes)
print('\nПропуски после обработки:')
print(clean.isna().sum())
````
